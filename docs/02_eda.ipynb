```python
from pathlib import Path

import pandas as pd

data_path = Path('../data/raw/uhpfrc_mix_design_master.csv')
if not data_path.exists():
    data_path = Path('data/raw/uhpfrc_mix_design_master.csv')

df = pd.read_csv(data_path)
df.head()
```

```python
df.info()
display(df.describe())
display(df.isnull().sum())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

numeric_columns = [
    'cement',
    'silica_fume',
    'fly_ash',
    'sand',
    'water',
    'superplasticizer',
    'fiber_content_percent',
    'water_binder_ratio',
    'curing_age_days',
    'curing_temp_celsius',
    'compressive_strength_MPa',
]
numeric_df = df[numeric_columns].apply(pd.to_numeric, errors='coerce')
correlation = numeric_df.corr()

if correlation.dropna(how='all').empty:
    print('No numeric rows available yet for a correlation heatmap.')
else:
    plt.figure(figsize=(12, 8))
    sns.heatmap(correlation, annot=True, cmap='coolwarm', center=0, fmt='.2f')
    plt.title('Correlation Heatmap of Numeric UHPFRC Mix Design Columns')
    plt.tight_layout()
    plt.show()
```

```python
hist_columns = [
    'cement',
    'fiber_content_percent',
    'water_binder_ratio',
    'compressive_strength_MPa',
]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for ax, column in zip(axes.ravel(), hist_columns):
    values = pd.to_numeric(df[column], errors='coerce').dropna()
    sns.histplot(values, kde=True, ax=ax)
    ax.set_title(column)

plt.tight_layout()
plt.show()
```
